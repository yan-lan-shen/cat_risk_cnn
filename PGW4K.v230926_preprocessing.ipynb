```python
from pathlib import Path
from scipy import sparse

import pandas as pd
import numpy as np
import xarray as xr
import glob
import os
```

```python
files_list = sorted(glob.glob("/lfs/......")) # a list of all historical file names
```

```python
xr.open_dataset(files_list[0])
```

```python
keep_vars = ["Times", "U", "V", "U10", "V10",]

output_dir = Path("/lfs/home/yanlan/climada/tc-risk/......")

for file in files_list:
    input_file = Path(file)

    # Retain the exact original filename
    output_file = output_dir/input_file.name

    with xr.open_dataset(input_file, decode_times=False) as ds:
        save_file = ds[keep_vars]

        assert "XLAT" in save_file.coords
        assert "XLONG" in save_file.coords
        
        save_file.to_netcdf(output_file)
```
